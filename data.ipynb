```python
import pandas as pd
import os
from vnstock import Quote

# --- 1. CẤU HÌNH ---
start_date = '2018-01-01'
end_date = '2025-12-31'

# Lưu vào thư mục tên là 'csv_data' nằm ngay trong folder dự án hiện tại
save_folder = './csv_data' 

# --- 2. HÀM TẢI DỮ LIỆU ---
def download_stock_to_csv(symbol, start, end, folder):
    print(f"🔄 Đang xử lý mã: {symbol}...")
    
    try:
        # Khởi tạo Quote (theo chuẩn Vnstock mới nhất)
        quote = Quote(symbol=symbol, source='VCI')
        df = quote.history(start=start, end=end, interval='1D')
        
        if df is not None and not df.empty:
            # Tạo thư mục nếu chưa có
            if not os.path.exists(folder):
                os.makedirs(folder)
            
            # Tạo đường dẫn file: ./csv_data/FPT.csv
            file_path = os.path.join(folder, f"{symbol}.csv")
            
            # Lưu file CSV
            df.to_csv(file_path, index=False, encoding='utf-8-sig')
            
            print(f"✅ Đã lưu file tại: {file_path}")
            print(f"📊 Số lượng dòng: {len(df)}")
            return file_path
        else:
            print(f"⚠️ Không có dữ liệu cho mã {symbol}")
            return None
            
    except Exception as e:
        print(f"❌ Lỗi: {e}")
        return None
```

```python
save_folder = './financial_data' 

# --- HÀM TẢI CHỈ SỐ TÀI CHÍNH ---
def download_financial_ratio(symbol, folder):
    print(f"📊 Đang tải chỉ số tài chính: {symbol}...", end=" ")
    
    try:
        # 1. Khởi tạo đối tượng Finance
        finance = Finance(symbol=symbol, source='VCI')
        
        # 2. Lấy dữ liệu (theo settings bạn yêu cầu: Quý, tiếng Anh, bỏ dòng trống)
        df = finance.ratio(period='quarter', lang='en', dropna=True)
        
        # 3. Kiểm tra dữ liệu
        if df is not None and not df.empty:
            # Tạo thư mục nếu chưa có
            if not os.path.exists(folder):
                os.makedirs(folder)
            
            # Thêm cột Ticker để dễ nhận biết
            df.insert(0, 'Ticker', symbol)
            
            # Tạo tên file: ACB_ratio.csv
            file_path = os.path.join(folder, f"{symbol}_ratio.csv")
            
            # Lưu file CSV
            df.to_csv(file_path, index=False, encoding='utf-8-sig')
            
            print(f"✅ OK ({len(df)} dòng)")
            return file_path
        else:
            print(f"⚠️ Không có dữ liệu.")
            return None
            
    except Exception as e:
        print(f"❌ Lỗi: {e}")
        return None
```

```python
import pandas as pd
import os

# --- 1. CẤU HÌNH ---
# Thư mục đầu vào (chứa dữ liệu thô/lỗi)
input_price_folder = '/workspaces/codespaces-jupyter/csv_data'
input_ratio_folder = '/workspaces/codespaces-jupyter/financial_data'

# Thư mục đầu ra (chứa dữ liệu sạch, chỉ từ năm 2018)
output_price_folder = '/workspaces/codespaces-jupyter/cleaned_price_data'
output_ratio_folder = '/workspaces/codespaces-jupyter/cleaned_ratio_data'

START_YEAR = 2017

# --- 2. CÁC HÀM HỖ TRỢ ĐỌC & SỬA LỖI HEADER (Dùng lại logic cũ) ---
def read_financial_csv_smart(filepath):
    """Đọc file tài chính, tự động bỏ qua các dòng Meta rác ở đầu"""
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            lines = [f.readline() for _ in range(10)] # Quét 10 dòng đầu
        
        header_idx = -1
        for i, line in enumerate(lines):
            # Tìm dòng chứa yearReport (dấu hiệu của header đúng)
            if 'yearReport' in line:
                header_idx = i
                break
        
        if header_idx != -1:
            return pd.read_csv(filepath, header=header_idx)
        return pd.read_csv(filepath)
    except Exception:
        return None

def clean_financial_columns(df):
    """Làm sạch tên cột (bỏ tiếng Việt dài dòng)"""
    new_mapper = {}
    prefixes = ["Chỉ tiêu cơ cấu nguồn vốn", "Chỉ tiêu khả năng sinh lợi", 
                "Chỉ tiêu thanh khoản", "Chỉ tiêu định giá", "Meta"]
    
    for col in df.columns:
        col_str = str(col).strip()
        # Giữ lại cột quan trọng
        if "yearReport" in col_str:
            new_mapper[col] = "yearReport"
            continue
        if "lengthReport" in col_str:
            new_mapper[col] = "lengthReport"
            continue
            
        # Xóa tiền tố
        clean_name = col_str
        for p in prefixes:
            clean_name = clean_name.replace(p, "")
        clean_name = clean_name.replace("\n", "").replace(".", "").strip()
        
        if clean_name != col_str:
            new_mapper[col] = clean_name
            
    if new_mapper:
        df = df.rename(columns=new_mapper)
    return df

# --- 3. HÀM CLEAN FILE GIÁ (PRICE) ---
def clean_price_file(filename):
    input_path = os.path.join(input_price_folder, filename)
    output_path = os.path.join(output_price_folder, filename)
    
    try:
        df = pd.read_csv(input_path)
        
        # Chuyển cột time sang datetime
        if 'time' not in df.columns:
            print(f"⚠️ {filename}: Thiếu cột 'time'. Bỏ qua.")
            return

        df['time'] = pd.to_datetime(df['time'])
        
        # LỌC DỮ LIỆU: Lấy từ năm 2018 trở đi
        df_clean = df[df['time'].dt.year >= START_YEAR]
        
        # Sắp xếp tăng dần theo thời gian
        df_clean = df_clean.sort_values('time')
        
        # Lưu file
        if not df_clean.empty:
            df_clean.to_csv(output_path, index=False)
            print(f"✅ Giá {filename}: Giữ lại {len(df_clean)} dòng (từ {START_YEAR})")
        else:
            print(f"⚠️ Giá {filename}: Không có dữ liệu sau năm {START_YEAR}")
            
    except Exception as e:
        print(f"❌ Lỗi file giá {filename}: {e}")

# --- 4. HÀM CLEAN FILE TÀI CHÍNH (RATIO) ---
def clean_ratio_file(filename):
    input_path = os.path.join(input_ratio_folder, filename)
    output_path = os.path.join(output_ratio_folder, filename)
    
    try:
        # 1. Đọc thông minh (xử lý dòng Meta)
        df = read_financial_csv_smart(input_path)
        if df is None: return

        # 2. Sửa tên cột
        df = clean_financial_columns(df)
        
        # 3. Kiểm tra cột bắt buộc
        if 'yearReport' not in df.columns:
            print(f"⚠️ Ratio {filename}: Không tìm thấy cột yearReport sau khi clean.")
            return

        # 4. LỌC DỮ LIỆU: yearReport >= 2018
        # Chuyển yearReport sang số (đề phòng bị hiểu là text)
        df['yearReport'] = pd.to_numeric(df['yearReport'], errors='coerce')
        df_clean = df[df['yearReport'] >= START_YEAR]
        
        # 5. Loại bỏ cột rác (Ticker, Unnamed...) để file sạch đẹp
        cols_to_drop = ['ticker', 'Ticker'] + [c for c in df_clean.columns if 'Unnamed' in str(c)]
        df_clean = df_clean.drop(columns=[c for c in cols_to_drop if c in df_clean.columns])
        
        # Lưu file
        if not df_clean.empty:
            df_clean.to_csv(output_path, index=False, encoding='utf-8-sig')
            print(f"✅ Ratio {filename}: Giữ lại {len(df_clean)} dòng (từ {START_YEAR})")
        else:
            print(f"⚠️ Ratio {filename}: Không có báo cáo nào từ năm {START_YEAR}")

    except Exception as e:
        print(f"❌ Lỗi file ratio {filename}: {e}")

# --- 5. THỰC THI ---
# Tạo thư mục đầu ra nếu chưa có
if not os.path.exists(output_price_folder): os.makedirs(output_price_folder)
if not os.path.exists(output_ratio_folder): os.makedirs(output_ratio_folder)

print(f"🚀 BẮT ĐẦU CLEAN DỮ LIỆU (Giữ lại từ năm {START_YEAR})...\n")

# A. Xử lý File Giá
print("--- Đang xử lý dữ liệu GIÁ ---")
price_files = [f for f in os.listdir(input_price_folder) if f.endswith('.csv')]
for f in price_files:
    clean_price_file(f)

print("\n--- Đang xử lý dữ liệu TÀI CHÍNH ---")
# B. Xử lý File Tài Chính
ratio_files = [f for f in os.listdir(input_ratio_folder) if f.endswith('.csv')]
for f in ratio_files:
    clean_ratio_file(f)

print(f"\n🎉 Hoàn tất! Dữ liệu sạch nằm tại:\n1. {output_price_folder}\n2. {output_ratio_folder}")
```

```python
import pandas as pd
import os
from pandas.tseries.offsets import MonthEnd, DateOffset

# --- 1. CẤU HÌNH ---
# Đường dẫn thư mục chứa dữ liệu đầu vào (Đã sạch)
price_folder = '/workspaces/codespaces-jupyter/cleaned_price_data'
ratio_folder = '/workspaces/codespaces-jupyter/cleaned_ratio_data'

# Đường dẫn thư mục lưu kết quả sau khi merge
output_folder = '/workspaces/codespaces-jupyter/final_merged_data'

# Cấu hình độ trễ và năm bắt đầu
SAFETY_LAG_DAYS = 45
START_YEAR = 2018

# --- 2. HÀM TÍNH NGÀY CÔNG BỐ ---
def calculate_available_date(row):
    """
    Hàm tính toán ngày dữ liệu tài chính thực sự có sẵn để giao dịch.
    Công thức: Ngày cuối cùng của quý + 45 ngày (SAFETY_LAG_DAYS).
    Ví dụ: Quý 1 kết thúc 31/03 -> Ngày có dữ liệu là 15/05.
    """
    try:
        year = int(row['yearReport'])
        quarter = int(row['lengthReport'])
        
        # Xác định tháng cuối của quý (3, 6, 9, 12)
        month = quarter * 3
        
        # Lấy ngày cuối cùng của tháng đó
        quarter_end_date = pd.to_datetime(f"{year}-{month}-01") + MonthEnd(0)
        
        # Cộng thêm số ngày trễ
        return quarter_end_date + DateOffset(days=SAFETY_LAG_DAYS)
    except:
        return pd.NaT

# --- 3. HÀM XỬ LÝ CHO MỘT MÃ CỔ PHIẾU ---
def process_single_ticker(ticker, price_path, ratio_path, output_path):
    try:
        # A. ĐỌC VÀ XỬ LÝ DỮ LIỆU GIÁ
        if not os.path.exists(price_path):
            print(f"[Canh bao] Khong tim thay file gia cho ma: {ticker}")
            return

        df_price = pd.read_csv(price_path)
        
        # Chuyển đổi cột time sang định dạng datetime
        df_price['time'] = pd.to_datetime(df_price['time'])
        
        # Lọc dữ liệu: Chỉ giữ lại các dòng giá từ năm 2018 trở đi
        # Sắp xếp tăng dần theo thời gian là bắt buộc để dùng merge_asof
        df_price = df_price[df_price['time'].dt.year >= START_YEAR].sort_values('time')
        
        if df_price.empty:
            print(f"[Canh bao] Ma {ticker}: Khong co du lieu gia tu nam {START_YEAR}.")
            return

        # B. ĐỌC VÀ XỬ LÝ DỮ LIỆU TÀI CHÍNH
        if not os.path.exists(ratio_path):
            print(f"[Canh bao] Ma {ticker}: Khong tim thay file chi so tai chinh.")
            return

        df_ratio = pd.read_csv(ratio_path)
        
        # Tạo cột 'trading_available_date' áp dụng độ trễ
        df_ratio['trading_available_date'] = df_ratio.apply(calculate_available_date, axis=1)
        
        # Loại bỏ các dòng không tính được ngày và sắp xếp theo ngày này
        df_ratio = df_ratio.dropna(subset=['trading_available_date']).sort_values('trading_available_date')
        
        # Lưu ý: Không lọc df_ratio theo năm 2018. 
        # Lý do: Để merge được cho ngày 01/01/2018, cần dữ liệu báo cáo của năm 2017.

        # C. THỰC HIỆN MERGE (GHEP DU LIEU)
        # merge_asof: Tìm bản ghi bên phải (ratio) gần nhất trong quá khứ so với bên trái (price)
        df_merged = pd.merge_asof(
            df_price,
            df_ratio,
            left_on='time',
            right_on='trading_available_date',
            direction='backward'
        )
        
        # D. LÀM SẠCH KẾT QUẢ
        # Loại bỏ các cột phụ trợ không cần thiết cho mô hình
        cols_drop = ['yearReport', 'lengthReport', 'trading_available_date', 'Unnamed: 0']
        # Chỉ drop những cột thực sự tồn tại trong dataframe
        cols_to_drop_actual = [c for c in cols_drop if c in df_merged.columns]
        df_merged = df_merged.drop(columns=cols_to_drop_actual)

        # Lưu file kết quả
        df_merged.to_csv(output_path, index=False)
        print(f"[Thanh cong] {ticker}: Da merge va luu {len(df_merged)} dong.")
        
    except Exception as e:
        print(f"[Loi] Xay ra loi voi ma {ticker}: {e}")

# --- 4. CHẠY VÒNG LẶP CHÍNH ---
# Tạo thư mục đầu ra nếu chưa tồn tại
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

print(f"Bat dau merge du lieu...")
print(f"Cau hinh: Lag {SAFETY_LAG_DAYS} ngay, lay du lieu tu nam {START_YEAR}")
print("-" * 30)

# Lấy danh sách file từ thư mục giá
# Giả sử tên file giá là "ACB.csv", ta sẽ lấy ticker là "ACB"
price_files = [f for f in os.listdir(price_folder) if f.endswith('.csv')]

count = 0
for p_file in price_files:
    ticker = p_file.replace('.csv', '') # Lấy tên mã
    
    # Đường dẫn file giá
    p_path = os.path.join(price_folder, p_file)
    
    # Xác định đường dẫn file tài chính tương ứng
    # Giả định file tài chính có tên dạng "ACB_ratio.csv"
    r_file = f"{ticker}_ratio.csv" 
    r_path = os.path.join(ratio_folder, r_file)
    
    # Đường dẫn file đầu ra
    out_path = os.path.join(output_folder, f"{ticker}_merged.csv")
    
    # Gọi hàm xử lý
    process_single_ticker(ticker, p_path, r_path, out_path)
    count += 1

print("-" * 30)
print(f"Hoan tat! Da xu ly {count} ma co phieu.")
print(f"Thu muc ket qua: {output_folder}")
```
