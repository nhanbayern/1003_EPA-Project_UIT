### XGBoost for Risk Premium Prediction

The risk premium prediction follows the general asset pricing model:

$$r_{i,t+1} = E_t(r_{i,t+1}) + \varepsilon_{i,t+1}$$

Where:
- $r_{i,t+1}$ = Actual return of asset $i$ at time $t+1$
- $E_t(r_{i,t+1})$ = Expected return of asset $i$ at time $t+1$ given information at time $t$
- $\varepsilon_{i,t+1}$ = Unexpected return (error term)
