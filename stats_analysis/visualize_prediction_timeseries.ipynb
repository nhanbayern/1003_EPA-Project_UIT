# Prediction Plotly Dashboard

Interactive dashboard for `true_volatility` vs `predict_volatility` diagnostics. This notebook loads `output/merged_predictions/merged_all_predictions_24_7.csv` and displays controls. Charts render only after clicking **Render** or **Show Outliers**; no PNGs or output folders are created.


```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

INPUT_RELATIVE = Path("output") / "merged_predictions" / "merged_all_predictions_24_7.csv"
PROJECT_ROOT = next(
    (path for path in [Path.cwd(), *Path.cwd().parents] if (path / INPUT_RELATIVE).exists()),
    Path.cwd(),
)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from stats_analysis import PredictionPlotlyDashboard

pd.set_option("display.max_columns", 120)
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:.6f}".format)

INPUT_CSV = PROJECT_ROOT / INPUT_RELATIVE
if not INPUT_CSV.exists():
    raise FileNotFoundError(f"Missing prediction CSV: {INPUT_CSV}")

dashboard = PredictionPlotlyDashboard.from_csv(
    INPUT_CSV,
    default_group="GARCH / No Tier",
    default_model="GARCH / No Tier / GARCH-LSTM-Hybrid",
    default_dataset="DAX_40",
    default_horizon=1,
)

print(f"Loaded prediction file: {INPUT_CSV}")
print(f"Rows: {len(dashboard.df):,}")
print(f"Datasets: {dashboard.df['dataset'].nunique()}")
print(f"Model labels: {dashboard.df['model_label'].nunique()}")
print(f"Horizons: {sorted(dashboard.df['horizon'].dropna().astype(int).unique().tolist())}")

display(dashboard.display())
```
