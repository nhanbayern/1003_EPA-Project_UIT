# Prediction Plotly Dashboard

Interactive dashboard for `true_volatility` vs `predict_volatility` diagnostics. This notebook only loads `output/merged_all_predictions.csv` and displays controls. Charts render only after clicking **Render** or **Show Outliers**; no PNGs or output folders are created.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "output" / "merged_all_predictions.csv").exists():
    PROJECT_ROOT = Path("D:/UIT_LEARNING_MATERIAL/07.Research/code")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from stats_analysis import PredictionPlotlyDashboard

pd.set_option("display.max_columns", 120)
pd.set_option("display.max_rows", 80)
pd.set_option("display.float_format", "{:.6f}".format)

INPUT_CSV = PROJECT_ROOT / "output" / "merged_all_predictions.csv"
if not INPUT_CSV.exists():
    raise FileNotFoundError(f"Missing prediction CSV: {INPUT_CSV}")

df = pd.read_csv(INPUT_CSV, low_memory=False)
dashboard = PredictionPlotlyDashboard(
    df,
    default_group="GARCH / No Tier",
    default_model="GARCH / No Tier / GARCH-LSTM-Hybrid",
    default_dataset="DAX_40",
    default_horizon=1,
)

display(dashboard.display())
```
